# Week 1 Assignment: Anscombe's Quartet Analysis
Analyze four datasets by calculating their mean and standard deviation using numpy, and understand what this tells us about the importance of data visualization.

```python
#import  libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load the Dataset
Load Anscombe's Quartet dataset from the datasets folder.

```python
# Load the Anscombe's Quartet dataset from the datasets folder.
df = pd.read_csv('../datasets/anscombes.csv')
dataset_ids = list(df.dataset.unique())
# Split into individual datasets

print("Dataset preview")
df.head()
```

# Calculate mean and standard daviation for each dataset
Calculate the mean and standard deviation for both x and y values in each of the four datasets using numpy.

```python
print("Statistical Properties of Anscombe's Quartet:\n")
print("=" * 70)

for dataset_id in dataset_ids: 
    dataset = df[df.dataset == dataset_id]
    x_values = dataset['x'].values
    y_values = dataset['y'].values

    x_mean = np.mean(x_values)
    x_std = np.std(x_values, ddof=1) 

    y_mean = np.mean(y_values)
    y_std = np.std(y_values, ddof=1)

    print(f"Dataset {dataset_id}:")
    print (f"  x - Mean: {x_mean:.2f}, Standard Deviation: {x_std:.2f}")
    print (f"  y - Mean: {y_mean:.2f}, Standard Deviation: {y_std:.2f}")
```

# Conclusion
Both the means and the Standard deviations for the x and y variables are exactly the same. 

Judiging from the plotting of the datasets u can see not all of the datasets are the same, yet they provide the same values. 

This means that the mean and standard deviation alone do NOT tell the full story. Visualization is essential top see the full picture.

# Create 4 plots for all of the datasets

```python
# Create scatter plots for all four datasets
fig, axes = plt.subplots(1, 4, figsize=(20, 5), sharex=True, sharey=True)

corrs = {}
for ax, dataset_id in zip(axes, dataset_ids):
    dataset = df[df['dataset'] == dataset_id]
    sns.scatterplot(ax=ax, x=dataset["x"], y=dataset["y"])
    ax.set_title(f'Dataset {dataset_id}')
    corrs[dataset_id] = np.corrcoef(dataset["x"], dataset["y"])[0, 1]

plt.tight_layout()
plt.show()

print("\nCorrelation coefficients:")
for dataset_id, corr in corrs.items():
    print(f"Dataset {dataset_id}: {corr:.3f}")
```
