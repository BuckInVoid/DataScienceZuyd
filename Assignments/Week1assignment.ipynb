# Week 1 Assignment: Anscombe's Quartet Analysis
Analyze four datasets by calculating their mean and standard deviation using numpy, and understand what this tells us about the importance of data visualization.

```python
#import  libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Anscombe
Load Anscombe's Quartet dataset from the datasets folder.

```python
# Load the Anscombe's Quartet dataset from the datasets folder.
df = pd.read_csv('../datasets/anscombes.csv')
dataset_ids = list(df.dataset.unique())
# Split into individual datasets

print("Dataset preview")
df.head()
```

# Calculate mean and standard daviation for each dataset
Calculate the mean and standard deviation for both x and y values in each of the four datasets using numpy.

```python
print("Statistical Properties of Anscombe's Quartet:\n")
print("=" * 70)

for dataset_id in dataset_ids: 
    dataset = df[df.dataset == dataset_id]
    x_values = dataset['x'].values
    y_values = dataset['y'].values

    x_mean = np.mean(x_values)
    x_std = np.std(x_values, ddof=1) 

    y_mean = np.mean(y_values)
    y_std = np.std(y_values, ddof=1)

    print(f"Dataset {dataset_id}:")
    print (f"  x - Mean: {x_mean:.2f}, Standard Deviation: {x_std:.2f}")
    print (f"  y - Mean: {y_mean:.2f}, Standard Deviation: {y_std:.2f}")
```

# Conclusion
Both the means and the Standard deviations for the x and y variables are exactly the same. 

Judiging from the plotting of the datasets u can see not all of the datasets are the same, yet they provide the same values. 

This means that the mean and standard deviation alone do NOT tell the full story. Visualization is essential top see the full picture.

# Create 4 plots for all of the datasets

```python
# Create scatter plots for all four datasets
fig, axes = plt.subplots(1, 4, figsize=(20, 5), sharex=True, sharey=True)

corrs = {}
for ax, dataset_id in zip(axes, dataset_ids):
    dataset = df[df['dataset'] == dataset_id]
    sns.scatterplot(ax=ax, x=dataset["x"], y=dataset["y"])
    ax.set_title(f'Dataset {dataset_id}')
    corrs[dataset_id] = np.corrcoef(dataset["x"], dataset["y"])[0, 1]

plt.tight_layout()
plt.show()

print("\nCorrelation coefficients:")
for dataset_id, corr in corrs.items():
    print(f"Dataset {dataset_id}: {corr:.3f}")
```

```python
# Correlation between lengte and schoenmaat
correlation = np.corrcoef(df['lengte'], df['schoenmaat'])[0, 1]
print(f"\nCorrelation coefficient: {correlation:.3f}")

# Scatter plot with regression line
plt.figure(figsize=(10, 6))
sns.regplot(x='lengte', y='schoenmaat', data=df, scatter_kws={'alpha':0.5})
plt.title(f'Height vs Shoe Size (r = {correlation:.3f})')
plt.xlabel('Lengte (cm)')
plt.ylabel('Schoenmaat')
plt.grid(alpha=0.3)
plt.show()
```

```python
# Side-by-side histograms
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(df['lengte'], bins=20, edgecolor='black', alpha=0.7)
axes[0].axvline(x_mean, color='red', linestyle='--', label=f'Mean: {x_mean:.1f}')
axes[0].set_title('Distribution of Lengte')
axes[0].set_xlabel('Lengte (cm)')
axes[0].set_ylabel('Frequency')
axes[0].legend()
axes[0].grid(alpha=0.3)

axes[1].hist(df['schoenmaat'], bins=20, edgecolor='black', alpha=0.7, color='orange')
axes[1].axvline(y_mean, color='red', linestyle='--', label=f'Mean: {y_mean:.1f}')
axes[1].set_title('Distribution of Schoenmaat')
axes[1].set_xlabel('Schoenmaat')
axes[1].set_ylabel('Frequency')
axes[1].legend()
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Box plots to spot outliers
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].boxplot(df['lengte'], vert=True)
axes[0].set_ylabel('Lengte (cm)')
axes[0].set_title('Lengte - Outlier Detection')
axes[0].grid(alpha=0.3)

axes[1].boxplot(df['schoenmaat'], vert=True)
axes[1].set_ylabel('Schoenmaat')
axes[1].set_title('Schoenmaat - Outlier Detection')
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

# Find outliers using IQR method
Q1_l, Q3_l = np.percentile(df['lengte'], [25, 75])
IQR_l = Q3_l - Q1_l
outliers_l = df[(df['lengte'] < Q1_l - 1.5*IQR_l) | (df['lengte'] > Q3_l + 1.5*IQR_l)]

Q1_s, Q3_s = np.percentile(df['schoenmaat'], [25, 75])
IQR_s = Q3_s - Q1_s
outliers_s = df[(df['schoenmaat'] < Q1_s - 1.5*IQR_s) | (df['schoenmaat'] > Q3_s + 1.5*IQR_s)]

print(f"\nOutliers in Lengte: {len(outliers_l)}")
print(f"Outliers in Schoenmaat: {len(outliers_s)}")
```
