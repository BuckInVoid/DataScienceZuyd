```python
#import  libraries

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset for Week 2 data understanding (use first column as index)
df_dc2 = pd.read_csv('../datasets/data_dc2.csv', index_col=0)
df_dc2.head()
```

```python
# Basic info and descriptive statistics
df_dc2.info()
df_dc2.describe(include='all')
```

```python
# Missing values and class distribution
# ...existing code...
# Missing values and class distribution
print('Missing values:', df_dc2.isnull().sum())
print('Class distribution:', df_dc2['class'].value_counts(dropna=False))
```

```python
# Simple plots: distributions and scatter by class
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
sns.histplot(df_dc2['x'], kde=True)
plt.title('x distribution')
plt.subplot(1,2,2)
sns.histplot(df_dc2['y'], kde=True)
plt.title('y distribution')
plt.tight_layout()
plt.show()
plt.figure(figsize=(6,6))
sns.scatterplot(data=df_dc2, x='x', y='y', hue='class', palette='Set1')
plt.title('x vs y by class')
plt.show()
```
